```python
plt.hist(hotel['lead_time'].dropna(), bins=30, color='orange', edgecolor='black')

plt.title('Histogram of Lead Time', fontsize=14, weight='bold')
plt.xlabel('Lead Time (Days)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)

plt.show()
```

```python
for col in hotel.columns:
# 1. Agar column Numerical hai (Histogram)
    if pd.api.types.is_numeric_dtype(hotel[col]):
        plt.figure(figsize=(7, 3.5))  # Compact size
        plt.hist(hotel[col], bins=30, color='orange', edgecolor='black')
        plt.title(f'Histogram: {col}', fontsize=11, weight='bold')
        plt.xlabel(col, fontsize=10)
        plt.ylabel('Frequency', fontsize=10)
        plt.tight_layout()
        plt.show()
```

```python
counts = hotel[col].value_counts()

plt.figure(figsize=(8, 3.5))
plt.bar(counts.index.astype(str), counts.values, color='skyblue', edgecolor='black', width = 0.6)
plt.title(f'Bar Plot: {col}', fontsize=11, weight='bold')
plt.xlabel(col, fontsize=10)
plt.ylabel('Count', fontsize=10)
plt.xticks(rotation=45, ha='right', fontsize=9)
plt.tight_layout()
plt.show()
```

# Adult- Data

```python
import pandas as pd
import numpy as np
```

```python
adult = pd.read_csv("/kaggle/input/datasets/rajahassan89/adult-file/adult.csv")
```

```python
(adult == "?").sum()
```

```python
(adult['occupation'] =="?")
```

```python
adult['native-country'] = adult['native-country'].replace("?", 'Unknown')
```

```python
adult['workclass'] = adult['workclass'].replace('?', adult['workclass'].mode()[0])
```

```python
adult['occupation'] = adult['occupation'].replace('?', adult['occupation'].mode()[0])
```

```python
adult['income'] = adult['income'].str.replace('K', '000' , regex = False)
```

```python
(adult['income'].value_counts())
```

```python
adult.head()
```

```python
import matplotlib.pyplot as plt
```

```python
for col in adult.columns:
    if pd.api.types.is_numeric_dtype(adult[col]):
        plt.figure(figsize=(7, 3.5))
        plt.hist(adult[col].dropna(), bins=30, color='orange', edgecolor='black')
        plt.title(f'Histogram: {col}', fontsize=11, weight='bold')
        plt.xlabel(col, fontsize=10)
        plt.ylabel('Frequency', fontsize=10)
        plt.tight_layout()
        plt.show()
```
